# Data merging
This script merged data on Orbis and Orbis IP separately for the given list of Italian firms.

```python
###  setup
import pandas as pd
import glob
 
# the two folders (Orbis = financials, Orbis IP = patents)
orbis_dir    = r"C:\Users\Popov\Documents\Studies\IMT_studies\Projects\Econ_2\Scopus\Initial_Data"
orbis_ip_dir = r"C:\Users\Popov\Documents\Studies\IMT_studies\Projects\Econ_2\Scopus_IP\Initial_Data"
```

### Financial data (Orbis)

```python
### stack the 3 financial files (strip broken styles first)
import zipfile, io

def read_orbis(path):
    # an xlsx is a zip; rebuild it with an empty styles.xml so openpyxl
    # doesn't choke on Orbis's stylesheet (the applyNumFmt bug)
    buf = io.BytesIO()
    with zipfile.ZipFile(path, "r") as zin:
        with zipfile.ZipFile(buf, "w") as zout:
            for item in zin.namelist():
                data = zin.read(item)
                if item == "xl/styles.xml":
                    # minimal valid empty stylesheet
                    data = (b'<?xml version="1.0" encoding="UTF-8" standalone="yes"?>'
                            b'<styleSheet xmlns="http://schemas.openxmlformats.org/'
                            b'spreadsheetml/2006/main"/>')
                zout.writestr(item, data)
    buf.seek(0)
    return pd.read_excel(buf, sheet_name="Results", engine="openpyxl")

fin_files = glob.glob(orbis_dir + r"\*.xlsx")
print(fin_files)

financials = pd.concat([read_orbis(f) for f in fin_files], ignore_index=True)
print(financials.shape) # correct shape
```

### Patents data (Orbis IP)

```python
### stack the 7 patent files (reuse read_orbis)
pat_files = glob.glob(orbis_ip_dir + r"\*.xlsx")
print(pat_files)

patents = pd.concat([read_orbis(f) for f in pat_files], ignore_index=True)
print(patents.shape)
```

```python
### drop USD columns from financials
usd_cols = [c for c in financials.columns if "USD" in str(c)]
print("dropping:", usd_cols)
financials = financials.drop(columns=usd_cols)
print(financials.shape)
```

```python
### fill the publication number down so every row is tagged to its patent
patents["Publication number"] = patents["Publication number"].ffill()
```

```python
## drop technical column
patents.drop(columns=[c for c in patents.columns if str(c).startswith("Unnamed")],
             inplace=True, errors="ignore")

## convert Excel date serials to real dates
for col in ["Priority date", "Grant date"]:
    patents[col] = pd.to_datetime(patents[col], unit="D", origin="1899-12-30")
patents.sample(3)
```

### Merging

```python
### make matching keys (string, upper, stripped)
# company side:
financials["bvd_key"] = financials["BvD ID number"].astype(str).str.strip().str.upper()
# patent side (the applicant is the company that filed it):
patents["bvd_key"] = patents["Applicant(s) BvD ID Number(s)"].astype(str).str.strip().str.upper()
```

```python
### SAVE CITATIONS before any filtering (lose nothing)
# each backward-citation value, tagged to its patent via publication number
citations = patents.loc[
    patents["Backward citations"].notna(),
    ["Publication number", "Backward citations"]
].copy()
print("citation links saved:", len(citations))
```

```python
### patent-level table: keep only rows that carry an applicant (can join to a firm)
patents_lvl = patents[patents["Applicant(s) BvD ID Number(s)"].notna()].copy()
print("patent rows with applicant:", len(patents_lvl))
```

```python
### company-centric merge (keep every company, attach its patents)
merged = financials.merge(patents_lvl, on="bvd_key", how="left")
merged.drop(columns=[c for c in merged.columns if str(c).startswith("Unnamed")],
            inplace=True, errors="ignore")
print("merged shape:", merged.shape)
```

```python
# how many companies matched at least one patent?
matched = merged.loc[merged["Publication number"].notna(), "bvd_key"].nunique()
total   = financials["bvd_key"].nunique()
print(f"{matched} of {total} companies have >=1 patent ({100*matched/total:.1f}%)")
```

## Export

```python
### clean column names for Stata, then export all four tables
def clean_cols(df):
    df = df.copy()
    cols = (pd.Index(df.columns).astype(str)
              .str.replace(r"[^0-9a-zA-Z]", "_", regex=True)
              .str.strip("_").str.lower())
    # Stata max var-name length is 32; truncate but keep years distinct
    cols = cols.str[:30]
    # de-duplicate any names that now collide (append _1, _2, ...)
    seen = {}
    out = []
    for c in cols:
        if c in seen:
            seen[c] += 1
            out.append(f"{c[:27]}_{seen[c]}")
        else:
            seen[c] = 0
            out.append(c)
    df.columns = out
    # Stata dislikes mixed object columns -> clean strings
    for c in df.columns:
        if df[c].dtype == object:
            df[c] = df[c].astype(str).replace({"nan": "", "None": "", "<NA>": ""})
    return df

export_dir = r"C:\Users\Popov\Documents\Studies\IMT_studies\Projects\Econ_2\Datasets"

clean_cols(financials ).to_stata(export_dir + r"\financials.dta", write_index=False, version=118)
clean_cols(patents_lvl).to_stata(export_dir + r"\patents.dta",    write_index=False, version=118)
clean_cols(citations  ).to_stata(export_dir + r"\citations.dta",  write_index=False, version=118)
clean_cols(merged     ).to_stata(export_dir + r"\merged.dta",     write_index=False, version=118)
print("exported: financials / patents / citations / merged  (.dta)")
```

### Owners of cited patents
Separate Orbis IP pull: publication number -> applicant BvD, for the patents our firms cite.

```python
cited_dir = r"C:\Users\Popov\Documents\Studies\IMT_studies\Projects\Econ_2\Scopus_IP\Cited"
cited_files = glob.glob(cited_dir + r"\*.xlsx")
print(cited_files)

cited_owners = pd.concat([read_orbis(f) for f in cited_files], ignore_index=True)
print(cited_owners.shape)

# junk column + continuation rows for multi-applicant patents
cited_owners.drop(columns=[c for c in cited_owners.columns if str(c).startswith("Unnamed")],
                  inplace=True, errors="ignore")
cited_owners["Publication number"] = cited_owners["Publication number"].ffill()

# find the applicant column whatever Orbis named it, build the key
appl_col = [c for c in cited_owners.columns if "bvd" in str(c).lower()][0]
print("using owner column:", appl_col)

cited_owners = cited_owners[cited_owners[appl_col].notna()].copy()
cited_owners["owner_key"] = cited_owners[appl_col].astype(str).str.strip().str.upper()

cited_owners = cited_owners[["Publication number", "owner_key"]].drop_duplicates()
print("pub-owner pairs:", len(cited_owners),
      "| distinct patents:", cited_owners["Publication number"].nunique())
```

```python
clean_cols(cited_owners).to_stata(export_dir + r"\cited_owners.dta",
                                  write_index=False, version=118)
print("saved cited_owners.dta")
```
